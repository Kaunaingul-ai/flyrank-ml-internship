## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

## My lane is a ranking/scoring task.

The goal is to rank content pages by how urgently they should be reviewed for possible decline. This is a ranking problem because the useful output is not only whether a page is declining, but which pages should be reviewed first. A ranked list can help an editor prioritize limited time and focus on the most important pages.

```python
task_type = "ranking/scoring"
decision = "Which content pages should be reviewed first?"

print("ML task:", task_type)
print("Decision supported:", decision)
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

##success metric: Precision@50

I would use Precision@50 because the goal is to give an editor a short ranked list of pages to review first. Precision@50 measures the proportion of the top 50 recommended pages that are actually declining.

For example, a Precision@50 of 0.80 would mean that 40 out of the top 50 recommended pages are truly declining. A higher Precision@50 is better.

```python
k = 50
correct_in_top_k = 40
example_precision_at_50 = correct_in_top_k / k

print("Success metric: Precision@50")
print("Example:", example_precision_at_50)
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

##Why ML can be more useful than one fixed rule

A fixed rule usually relies on only one or two conditions, such as low CTR or old content. In this dataset, page decline may be associated with several signals at the same time, including impressions, CTR, ranking position, content age, and update history.

A machine learning model can combine these signals and learn interactions between them. This can make it more useful for ranking pages than relying on one fixed threshold alone.

```python
check = df.copy()
check["is_declining"] = (check["trend_direction"] == "down").astype(int)

features_to_check = [
    "impressions_90d",
    "ctr",
    "avg_position",
    "content_age_days",
    "days_since_last_update",
    "word_count"
]

check.groupby("is_declining")[features_to_check].median().round(2)
```
